Smokers are the biggest cost driver in the dataset. Non-smokers’ medical charges stay relatively low and rise steadily with age, while smokers form a much higher, heavily skewed group whose costs often triple or quadruple those of non-smokers of the same age. When BMI is added, the difference becomes even sharper: for non-smokers, higher BMI only modestly increases costs (mainly after the obesity threshold of 30), but for smokers a high BMI multiplies charges dramatically, producing the highest values in the data (above $63,000).
Other patterns are milder. The dataset is skewed younger, with many people in their late teens and twenties (mostly non-smokers), which keeps overall average costs lower until age effects appear. Regional differences are small, though the southeast tends to have slightly higher average charges, partly because of more higher-risk individuals. Having children or dependents adds only a minor increase in costs compared with the much stronger effects of age and smoking.

Pre Processing

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(columns="charges")
y=df["charges"]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
from sklearn.preprocessing import LabelEncoder
binary_cols = ["sex", "smoker"]
for col in binary_cols:
  le = LabelEncoder()
  X_train[col] = le.fit_transform(X_train[col])
  X_test[col] = le.transform(X_test[col])

X_train=pd.get_dummies(X_train,columns=["region"],drop_first=True,dtype=int)
X_test=pd.get_dummies(X_test,columns=["region"],drop_first=True,dtype=int)
```

```python
X_train.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
num_cols=["age","bmi","children"]
scaler_X=MinMaxScaler()
X_train[num_cols]=scaler_X.fit_transform(X_train[num_cols])
X_test[num_cols]=scaler_X.transform(X_test[num_cols])

scaler_y = MinMaxScaler()
y_train=scaler_y.fit_transform(y_train.values.reshape(-1, 1))
y_test=scaler_y.transform(y_test.values.reshape(-1, 1))
```

Model

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(X_train,y_train)
y_pred=regressor.predict(X_test)

print("Type:",type(y_pred))
print("Size:",len(y_pred))
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae=mean_absolute_error(y_test, y_pred)
rmse=np.sqrt(mean_squared_error(y_test, y_pred))
r2=r2_score(y_test, y_pred)

print(f"\nMean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

Implementing Linear reg on our own

```python
class LinearRegression() :

    def __init__( self, learning_rate, iterations ) :
        self.learning_rate = learning_rate
        self.iterations = iterations
        
    # Function for model training
    def fit(self,X,Y):

        X=np.array(X)
        Y=np.array(Y).ravel()

        # no_of_training_examples, no_of_features
        self.m,self.n=X.shape

        self.W=np.zeros(self.n)

        self.b =0 # set this equal to 0
        self.X = X  # set this equal to X
        self.Y = Y # set this equal to Y
        
        # gradient descent learning
        for i in range(self.iterations) : # complete the range function
            self.update_weights()

        return self

    # Helper function to update weights in gradient descent
    def update_weights(self) :

        Y_pred=self.predict(self.X) # complete this line

        # calculate gradients
        dW = (2/self.m)*np.dot(self.X.T,(Y_pred-self.Y))# dL/dW -> L = MSE = (y -ypred)^2/m
        db = (2/self.m)*np.sum(Y_pred-self.Y)

        # write the code to update the weights
        self.W =self.W-self.learning_rate*dW
        self.b =self.b-self.learning_rate*db

        return self


    def predict(self,X) :

        return X.dot(self.W) + self.b
```

```python
# Create and train the custom Linear Regression model
model = LinearRegression(
    learning_rate=0.0000001,
    iterations=100000
)

model.fit(X_train, y_train)

# Make predictions
y_predd = model.predict(X_test)

# Check whether predictions contain NaN
print("NaN in predictions:", np.isnan(y_predd).any())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_predd)
rmse = np.sqrt(mean_squared_error(y_test, y_predd))
r2 = r2_score(y_test, y_predd)

print(f"\nMean Absolute Error (MAE): {mae:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"R² Score: {r2:.4f}")
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

degrees = [1, 2, 3, 4, 5]

results = []

for degree in degrees:

    # Create polynomial features
    poly = PolynomialFeatures(degree=degree, include_bias=False)

    X_train_poly = poly.fit_transform(X_train)
    X_test_poly = poly.transform(X_test)
    
    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    train_mse = mean_squared_error(y_train, y_train_pred)
    test_mse = mean_squared_error(y_test, y_test_pred)

    train_r2 = r2_score(y_train, y_train_pred)
    test_r2 = r2_score(y_test, y_test_pred)

    results.append({"Degree": degree,"Train MSE": train_mse,"Test MSE": test_mse,"Train R2": train_r2,"Test R2": test_r2})
results
```

```python
import pandas as pd

results_df = pd.DataFrame(results)

print(results_df)
```
